```python
import os
import yt_dlp
```

### USER INPUT

```python
playlist_url = input("Enter playlist URL: ").strip()

start_num = input("Download starting from video number (default = 1): ").strip()
end_num = input("Download up to which number? (press Enter for last): ").strip()
```

### Defaults

```python
start_num = int(start_num) if start_num else 1
end_num = int(end_num) if end_num else None

print(f"\nDownloading from #{start_num} to #{end_num if end_num else 'LAST'}\n")
```

### DOWNLOAD PATH

```python
download_path = "/content/playlist_downloads"
os.makedirs(download_path, exist_ok=True)
```

### YT-DLP OPTIONS:

```python
ydl_opts = {
    'outtmpl': os.path.join(download_path, '%(playlist_index)s - %(title)s.%(ext)s'),
    'format': 'bestvideo+bestaudio/best',
    'merge_output_format': 'mp4',
    'ignoreerrors': True,
    'playliststart': start_num,
}

if end_num:
    ydl_opts['playlistend'] = end_num

with yt_dlp.YoutubeDL(ydl_opts) as ydl:
    ydl.download([playlist_url])

print("\nDownload completed.")
print(f"Files saved in: {download_path}")
```

### ZIP full playlist

```python
import shutil
from google.colab import files

folder_path = "/content/playlist_downloads"
zip_name = "playlist_downloads"

# Create zip file
shutil.make_archive(zip_name, 'zip', folder_path)

# Download to device
files.download(f"{zip_name}.zip")
```
